```python
import pandas as pd
from pathlib import Path
```

De data die wordt gebruikt in deze verkenfase is de transcript van de video over de workshop over Breinschade van Sara. Deze video betreft niet een workshop gegeven in een klaslokaal, maar een op een met Tim. Dit is belangrijk om in het achterhoofd te houden met betrekking tot de kwaliteit van het transcript. Voor de volgende verkenning zal een passender voorbeeld worden genomen (qua audio), maar voor de eerste verkenfase is dit prima. 

```python
transcript_path = Path("../../Data-local/raw/Breinschade_Uitleg_Sara.txt")

with open(transcript_path, "r", encoding="utf-8") as f:
    transcript = f.read()

print(transcript[:2000])
```

Zoals te zien is, zijn de sprekers niet goed terug te vinden en is er wel sprake van timestamps. Voor dit specifieke voorbeeld is er gebruikgemaakt van Youtube-transcript.io, een online tool waarmee je de transcript van een video als txt bestand kunt downloaden. Ik ga nog onderzoek doen naar tools die o.a. ook de sprekers kan differentieren. 

```python
print("Aantal tekens:", len(transcript))
print("Aantal woorden:", len(transcript.split()))
print("Aantal vraagtekens:", transcript.count("?"))
```

```python
regels = transcript.splitlines()

vragen = [regel for regel in regels if "?" in regel]

for vraag in vragen[:20]:
    print(vraag)
```

```python
for i, vraag in enumerate(vragen[:30], start=1):
    print(f"{i}. {vraag}")
```
